## ACTIVIDAD A
Por medio de la librería ‘Pandas’, lea la base de datos, asigne nombres para las columnas, realice una imputación para la variable ‘Bare Nuclei’ por medio de la mediana y cambie las categorías para la variable ‘Class’ de 2 o 4 por 0 o 1

```python
# Importar librerias
import pandas as pd

# Cargar datos desde el repositorio
url = 'https://archive.ics.uci.edu/static/public/15/data.csv'
df = pd.read_csv(url)
df.head()
```

```python
# imputación para la variable ‘Bare Nuclei’ por medio de la mediana
med = df['Bare_nuclei'].median()
df['Bare_nuclei']=df['Bare_nuclei'].fillna(med)

# Nulos en Bare_nuclei
print("Nulos en Bare_nuclei:", df['Bare_nuclei'].isnull().sum())
```

```python
# cambiar las categorías para la variable ‘Class’ de 2 o 4 por 0 o 1
df['Class'] = df['Class'].replace({2: 0, 4: 1})
# Verificar resultados
print("Valores de 'Class':", df['Class'].unique())
```

## ACTIVIDAD B
Realice una detección de valores atípicos univariados por medio del método del rango intercuartílico con 3 de longitud a la derecha y 3 a la izquierda, y elimínelos. Además, realice una detección de valores atípicos multivariados por medio de las distancias de Mahalanobis y elimine aquellos valores que superen el valor de 30.

### Eliminación de valores atípicos univariados

```python
# Creamos una copia de la data
df1=df

# Detectar y eliminar outliers univariados con IQR * 3
for col in df1.select_dtypes(include='number').columns:
    Q1 = df1[col].quantile(0.25)
    Q3 = df1[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR
    df1 = df1[(df1[col] >= lower_bound) & (df1[col] <= upper_bound)]
```

```python
# Metricas antes de la eliminación de outliers
df.describe()
```

```python
# Métricas luego de la eliminación de outliers
df1.describe()
```


### Eliminación de outliers multivariados usando mahalanobis


```python
# Importamoslibrerias n
import numpy as np
from scipy.spatial.distance import mahalanobis
from numpy.linalg import pinv
```

```python
# Variables numéricas
num_vars = df1.select_dtypes(include='number')

# Convertir a matriz
X = num_vars.values
```

```python
# Media y matriz de covarianza inversa
mean_vec = np.mean(X, axis=0)
cov_matrix = np.cov(X, rowvar=False)
inv_cov_matrix = pinv(cov_matrix)
```

```python
# Calcular distancias de Mahalanobis
mahal_dist = [mahalanobis(x, mean_vec, inv_cov_matrix) for x in X]
```

```python
# Filtrar filas con distancia ≤ 30
df1['mahal_dist'] = mahal_dist
df1 = df1[df1['mahal_dist'] <= 30].drop(columns='mahal_dist')
```
